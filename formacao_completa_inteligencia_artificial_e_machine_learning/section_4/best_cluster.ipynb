```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score
from sklearn import datasets
```

```python
def compare_algorithms(X, max_clusters):
  results = []
  cluster_range = range(2, max_clusters + 1)
  for n_clusters in cluster_range:
    kmeans = KMeans(n_clusters=n_clusters, random_state=0, n_init='auto')
    agg = AgglomerativeClustering(n_clusters=n_clusters)
    eps = np.arange(0.1,0.9,0.1)

    for eps in eps:
      dbscan = DBSCAN(eps=eps, min_samples=5)
      clusters = dbscan.fit_predict(X)
      if len(set(clusters)) > 1:
        dbscan_score = silhouette_score(X, clusters)
        results.append(('DBSCAN', eps, dbscan_score))


    kmeans.fit_predict(X)
    agg.fit_predict(X)

    kmeans_score = silhouette_score(X, kmeans.labels_)
    agg_score = silhouette_score(X, agg.labels_)

    results.append(('KMeans', n_clusters, kmeans_score))
    results.append(('Agglomerative', n_clusters, agg_score))
    
  return results
```

```python
import h2o
from h2o.automl import H2OAutoML
```

```python
h2o.init(ip="localhost", port="8080")
```

```python
data = h2o.import_file("insurance.csv")
```

```python
data
```

```python
data = data.drop("C1")
```

```python
data
```

```python
train, test = data.split_frame(ratios=[0.7], seed=0)
```

```python
train
```

```python
model_automl = H2OAutoML(max_runtime_secs=180, seed=0, sort_metric="AUTO")
model_automl.train(y="Accident", training_frame=train)
```

```python
ranking = model_automl.leaderboard
ranking = ranking.as_data_frame()
ranking
```

```python
predict = model_automl.leader.predict(test)
predict
```

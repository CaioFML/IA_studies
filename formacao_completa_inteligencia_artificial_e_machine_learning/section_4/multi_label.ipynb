```python
from skmultilearn.adapt import MLARAM
from skmultilearn.problem_transform import BinaryRelevance, ClassifierChain, LabelPowerset
from sklearn.svm import SVC

from sklearn.model_selection import train_test_split
from sklearn.metrics import hamming_loss

import numpy as np
import pandas as pd
```

```python
music = pd.read_csv('musica.csv')
music.shape
```

```python
music.head()
```

```python
classe = music.iloc[:, 0:6].values
forecasters = music.iloc[:, 7:78].values
classe
```

```python
X_train, X_test, y_train, y_test = train_test_split(forecasters, classe, test_size=0.3, random_state=0)
```

```python
# ann = MLARAM()
# ann.fit(X_train, y_train)
```

```python
# prediction = ann.predict(X_test)
# print('Hamming Loss: ', hamming_loss(y_test, prediction))
# 0.24
```

```python
binary = BinaryRelevance(classifier=SVC())
binary.fit(X_train, y_train)
prediction = binary.predict(X_test)
print('Hamming Loss: ', hamming_loss(y_test, prediction))
```

```python
label = LabelPowerset(classifier=SVC())
label.fit(X_train, y_train)
prediction = label.predict(X_test)
print('Hamming Loss: ', hamming_loss(y_test, prediction))
```

```python
chain = ClassifierChain(classifier=SVC())
chain.fit(X_train, y_train)
prediction = chain.predict(X_test)
print('Hamming Loss: ', hamming_loss(y_test, prediction))
```

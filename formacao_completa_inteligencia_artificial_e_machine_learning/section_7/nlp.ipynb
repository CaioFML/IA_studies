```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer, SnowballStemmer, LancasterStemmer
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.tag import pos_tag, pos_tag_sents
import string
```

```python
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('words')
nltk.download('maxent_ne_chunker')
nltk.download('tagsets')
```

```python
Text = '''Nós somos feitos de poeira de estrelas. Nós somos uma maneira de o
cosmos se autoconhecer. A imaginação nos leva a mundos que nunca
 sequer existiram. Mas sem ela não vamos a lugar algum.  '''
print(Text)
```

```python
sentence = sent_tokenize(Text, language='portuguese')
print(type(sentence))
print(sentence)
```

```python
print(len(sentence))
```

```python
tokens = word_tokenize(Text, language='portuguese')
print(type(tokens))
print(tokens)
```

```python
stops = stopwords.words('portuguese')
print(stops)
```

```python
words_without_stopwords = [p for p in tokens if p not in stops]
print(words_without_stopwords)
```

```python
print(len(tokens), "  ", len(words_without_stopwords))
```

```python
print(string.punctuation)
```

```python
words_without_punctuation = [p for p in words_without_stopwords if p not in string.punctuation]
print(words_without_punctuation)
```

```python
frequency = nltk.FreqDist(words_without_punctuation)
frequency
```

```python
common = frequency.most_common(2)
common
```
